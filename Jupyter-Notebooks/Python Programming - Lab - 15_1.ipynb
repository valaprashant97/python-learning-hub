### 03) WAP to create Circle class with area and perimeter function to find area and perimeter of circle.

```python
class Circle:
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 3.14 * self.radius * self.radius

    def perimeter(self):
        return 2 * 3.14 * self.radius

r = float(input("Enter radius: "))
c = Circle(r)

print("Area:", c.area())
print("Perimeter:", c.perimeter())
```

### 05) Create a bank account class with methods to deposit, withdraw, and check balance.

```python
class BankAccount:
    def __init__(self, name, balance):
        self.name = name
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        print("Deposited:", amount)

    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
            print("Withdrawn:", amount)
        else:
            print("Insufficient Balance")

    def check_balance(self):
        print("Current Balance:", self.balance)

name = input("Enter a Name :")
balance = float(input("Enter a Balance :"))
acc = BankAccount(name, balance)

acc.deposit(2000)
acc.withdraw(3000)
acc.check_balance()
```

### 06) Create a class for managing inventory that includes attributes such as item name, price, quantity, and methods to add, remove, and update items.

```python
class Inventory:
    def __init__(self, name, price, quantity):
        self.name = name
        self.price = price
        self.quantity = quantity

    def add_item(self, qty):
        self.quantity += qty
        print("Item added. New quantity:", self.quantity)

    def remove_item(self, qty):
        if qty <= self.quantity:
            self.quantity -= qty
            print("Item removed. New quantity:", self.quantity)
        else:
            print("Not enough stock")

    def update_price(self, new_price):
        self.price = new_price
        print("Price updated :", self.price)

    def display(self):
        print("Item:", self.name)
        print("Price:", self.price)
        print("Quantity:", self.quantity)

name = input("Enter a Name of Product :")
price = float(input("Enter a Price :"))
quantity = float(input("Enter a Quantity :"))
item = Inventory(name, price, quantity)

item.display()
item.add_item(50)
item.remove_item(30)
item.update_price(12)
item.display()
```

### 09) Define Time class with hour and minute as data member. Also define addition method to add two time objects.

```python
class Time:
    def __init__(self, hour, minute):
        self.hour = hour
        self.minute = minute

    def add_time(self, t2):
        total_min = self.minute + t2.minute
        total_hour = self.hour + t2.hour

        if total_min >= 60:
            total_hour += total_min // 60
            total_min = total_min % 60

        return Time(total_hour, total_min)

    def display(self):
        print(f"{self.hour} hour {self.minute} minute")

hour1 = int(input("Enter a Fist hour : "))
minute1 = int(input("Enter a Fist hour : "))
t1 = Time(hour1, minute1)
hour2 = int(input("Enter a Secound hour : "))
minute2 = int(input("Enter a Secound hour : "))
t2 = Time(hour2, minute2)

t3 = t1.add_time(t2)

print("Time 1:")
t1.display()

print("Time 2:")
t2.display()

print("After Addition:")
t3.display()
```
