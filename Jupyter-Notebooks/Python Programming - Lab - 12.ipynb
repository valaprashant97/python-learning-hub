# Exception Handling

### 01) WAP to handle following exceptions:
1. ZeroDivisionError
2. ValueError
3. TypeError
#### Note: handle them using separate except blocks and also using single except block too.

```python
try:
    a = int(input("Enter first number: "))
    b = int(input("Enter second number: "))
    
    result = a / b
    print("Result: ", result)

except ZeroDivisionError:
    print("Division by zero is not allowed.")

except ValueError:
    print("Please enter valid integers.")

except TypeError:
    print("Invalid type used in operation.")
```

```python
a = int(input("Enter a number : "))
b = int(input("Enter a number : "))

try:
    ans = a/b
    print(ans)
except ZeroDivisionError:
    print("Can't devide by zero")
else:
    print("no errors occurred in the try block")
finally:
    print("This code always runs")
```

### 06) WAP to create an udf divide(a,b) that handles ZeroDivisionError.

```python
def divide(a,b):
    try:
        ans = a/b
        return ans
    except ZeroDivisionError:
        return "Can't devide by zero"
    
result = divide(20,0)
print(result);
```

### 07) WAP to accept item prices and calculate total. Raise ValueError if price is negative, otherwise print the total bill amount.

```python
total_bill = 0

try:
    a = input("Enter the price: ")
    price = float(a)
    
    if price < 0:
        raise ValueError("Price cannot be negative.")
    
    total_bill += price
    print(f"Total bill amount: {total_bill}")

except ValueError as e:
    print(f"Error: {e}")
except Exception as e:
    print(f"An unexpected error occurred: {e}")
```

### 08) Create a short program that prompts the user for a list of grades separated by commas. 
#### Split the string into individual grades and use a list comprehension to convert each string to an integer. 
#### You should use a try statement to inform the user when the values they entered cannot be converted.

```python
grades = input("Enter grades separated by commas: ")

try:
    grade_list = [int(g) for g in grades.split(",")]
    print("Grades:", grade_list)
except:
    print("Invalid input. Please enter only numbers separated by commas.")
```

### 09) Accept 5 subject marks (0-100). Use assert for validation.

#### Sample Input (Error Case)
#####  Enter marks: 110 

#### Sample Output (Error Case)
##### AssertionError: Marks should be between 0 and 100.

#### Sample Input (Normal Case)
##### Valid marks entered

#### Sample Output (Normal Case)
##### Average Marks displayed

```python
try:
    marks = []
    for i in range(1, 6):
        mark = float(input("Enter a marks : "))
        assert 0 <= mark <= 100, "marks should me 0 to 100"
        marks.append(mark)
    
    avg = sum(marks) / len(marks)
    print(avg)

except AssertionError as e:
    print(e)
except ValueError:
    print("Enter a numbers only")
```

### 11) WAP to raise your custom Exception named NegativeNumberError with the error message : "Cannot calculate the square root of a negative number" :
#### if the given number is negative.
#### otherwise print the square root of the given number.

```python
try:
    a = input("Enter the price: ")
    num = float(a)
    
    if num < 0:
        raise ValueError("Cannot calculate the square root of a negative number.")
    
    square = num*num
    print(square)
    
except ValueError as e:
    print(f"Error: {e}")
except Exception as e:
    print(f"An unexpected error occurred: {e}")
```
