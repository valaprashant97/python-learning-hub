### 10) Calculate area of a ractangle using object as an argument to a method.

```python
class Rectangle:
    def __init__(self, l, w):
        self.length = l
        self.width = w

class Area:
    def calculate(self, obj):
        return obj.length * obj.width

l = float(input("Enter length : "))
w = float(input("Enter width : "))

r = Rectangle(l, w)
a = Area()

print("Area of rectangle:", a.calculate(r))
```

### 11) Calculate the area of a square. 
### Include a Constructor, a method to calculate area named area() and a method named output() that prints the output and is invoked by area().

```python
class Square:
    def __init__(self, side):
        self.side = side

    def area(self):
        result = self.side * self.side
        self.output(result)

    def output(self, result):
        print("Area of square:", result)

s = float(input("Enter side: "))
sq = Square(s)
sq.area()
```

### 12) Calculate the area of a rectangle. 
### Include a Constructor, a method to calculate area named area() and a method named output() that prints the output and is invoked by area().
### Also define a class method that compares the two sides of reactangle. An object is instantiated only if the two sides are different; otherwise a message should be displayed : THIS IS SQUARE.

```python
class Rectangle:
    def __init__(self, l, w):
        self.length = l
        self.width = w

    @classmethod
    def create(cls, l, w):
        if l == w:
            print("THIS IS SQUARE")
        else:
            cls(l, w).area()

    def area(self):
        result = self.length * self.width
        self.output(result)

    def output(self, result):
        print("Area of rectangle:", result)


l = float(input("Enter length: "))
w = float(input("Enter width: "))

Rectangle.create(l, w)
```

### 13) Define a class Square having a private attribute "side". 
### Implement get_side and set_side methods to accees the private attribute from outside of the class.

```python
class Square:
    def __init__(self, side):
        self.__side = side

    def get_side(self):
        return self.__side

    def set_side(self, side):
        self.__side = side


s = Square(5)
print("Side:", s.get_side())
s.set_side(10)
print("Updated Side:", s.get_side())
```

### 14) Create a class Profit that has a method named getProfit that accepts profit from the user.
### Create a class Loss that has a method named getLoss that accepts loss from the user.
### Create a class BalanceSheet that inherits from both classes Profit and Loss and calculates the balanace. It has two methods getBalance() and printBalance().

```python
class Profit:
    def getProfit(self):
        self.profit = float(input("Enter profit: "))

class Loss:
    def getLoss(self):
        self.loss = float(input("Enter loss: "))

class BalanceSheet(Profit, Loss):
    def getBalance(self):
        self.balance = self.profit - self.loss

    def printBalance(self):
        print("Balance:", self.balance)

b = BalanceSheet()
b.getProfit()
b.getLoss()
b.getBalance()
b.printBalance()
```

### 15) WAP to demonstrate all types of inheritance.

```python
# Single Inheritance
class A:
    def showA(self):
        print("Class A")

class B(A):
    def showB(self):
        print("Class B")


# Multiple Inheritance
class C:
    def showC(self):
        print("Class C")

class D(A, C):
    def showD(self):
        print("Class D")


# Multilevel Inheritance
class E(B):
    def showE(self):
        print("Class E")


# Hierarchical Inheritance
class F(A):
    def showF(self):
        print("Class F")

class G(A):
    def showG(self):
        print("Class G")


# Hybrid Inheritance (combination)
class H(D, E):
    def showH(self):
        print("Class H")


# Objects
b = B()
b.showA()
b.showB()

d = D()
d.showA()
d.showC()
d.showD()

e = E()
e.showA()
e.showB()
e.showE()

f = F()
f.showA()
f.showF()

g = G()
g.showA()
g.showG()

h = H()
h.showA()
h.showC()
h.showB()
h.showH()
```

### 16) Create a Person class with a constructor that takes two arguments name and age. 
### Create a child class Employee that inherits from Person and adds a new attribute salary. 
### Override the __init__ method in Employee to call the parent class's __init__ method using the super() and then initialize the salary attribute.

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age


class Employee(Person):
    def __init__(self, name, age, salary):
        super().__init__(name, age)
        self.salary = salary


name = input("Enter name: ")
age = int(input("Enter age: "))
salary = float(input("Enter salary: "))

e = Employee(name, age, salary)

print("Name:", e.name)
print("Age:", e.age)
print("Salary:", e.salary)
```

### 17) Create a Shape class with a draw method that is not implemented. 
### Create three child classes Rectangle, Circle, and Triangle that implement the draw method with their respective drawing behaviors. 
### Create a list of Shape objects that includes one instance of each child class, and then iterate through the list and call the draw method on each object.

```python
class Shape:
    def draw(self):
        pass


class Rectangle(Shape):
    def draw(self):
        print("Drawing Rectangle")


class Circle(Shape):
    def draw(self):
        print("Drawing Circle")


class Triangle(Shape):
    def draw(self):
        print("Drawing Triangle")


shapes = [Rectangle(), Circle(), Triangle()]

for s in shapes:
    s.draw()
```
