### 03) Write a function that returns 1 if the given number is Prime or 0 otherwise.

```python
def isprime(n):
    for i in range(2,n):
        if(n%i==0):
            return 0;
        else:
            return 1;

prime= isprime(5)
print("Number is:",prime)
```

### 04) Write a function that returns the list of Prime numbers between given two numbers.

```python
def primes_between(a, b):
    primes = []
    for n in range(a, b + 1): 
        if n < 2:
            continue
        is_prime = True
        for i in range(2, n):
            if n % i == 0:
                is_prime = False
                break
        if is_prime:
            primes.append(n)
    return primes

primes_between(3,15)
```

### 05) Write a function that returns True if the given string is Palindrome or False otherwise.

```python
str=input("Enter the String:")
def ispalidrome(str):
    if(str==str[::-1]):
        return "true";
    else:
        return "false";

ispalidrome(str)
```

### 08) Write a recursive function to find nth term of Fibonacci Series.

```python
def fibonacci(n):
    prev = 0
    curr = 1

    for i in range(n):
        print(prev, end=" ")
        next = prev + curr
        prev = curr
        curr = next

n = int(input("Enter the N: "))
fibonacci(n)
```

### 11) Write a function to invert a given Dictionary.
#### hint: keys to values & values to keys
#### Before : {'a': 10, 'b':20, 'c':30, 'd':40}
#### After : {10:'a', 20:'b', 30:'c', 40:'d'}

```python
Before={'a': 10, 'b':20, 'c':30, 'd':40}

def invert(n):
    new={}
    for k,v in Before.items():
      new[v]=k
    return new
ans=invert(Before)
print("After:",ans)
```

### 12) Write a function that returns the number of uppercase and lowercase letters in the given string. 
#### example : Input : s1 = AbcDEfgh ,Ouptput : no_upper = 3, no_lower = 5

```python
s1 = "AbcDEfgh"
def count_case(s):
    no_upper = 0
    no_lower = 0
    for ch in s:
        if ch.isupper():
            no_upper += 1
        elif ch.islower():
            no_lower += 1
    return no_upper, no_lower

upper, lower = count_case(s1)
print(f"no_upper = {upper}, no_lower = {lower}")
```

### 16) Write udfs to call the functions with following types of arguments:
1. Positional Arguments
2. Keyword Arguments
3. Default Arguments
4. Variable Legngth Positional(*args) & variable length Keyword Arguments (**kwargs)
5. Keyword-Only & Positional Only Arguments

```python
def positional_arguments(a, b):
    return a + b
positional_arguments(5, 10)

def keyword_arguments(a=0, b=0):
    return a * b
keyword_arguments(b=4, a=3)

def default_arguments(a, b=2):
    return a ** b
default_arguments(3)

def variable_length_arguments(*args, **kwargs):
    return args, kwargs
variable_length_arguments(1, 2, 3, name="Alice", age=25)

def keyword_only_arguments(a, /, b, *, c):
    return a + b + c
keyword_only_arguments(1, b=2, c=3)
```
